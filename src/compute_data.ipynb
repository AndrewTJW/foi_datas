## 1. Congressional Research Service → physical supply exposure

The Congressional Research Service report describes the Strait of Hormuz's role in global energy flows. The flow charts provide context for the scale and composition of oil moving through this chokepoint.

**Source:** Congressional Research Service, "The Strait of Hormuz: Security Developments and Impacts on Oil, Gas, and Other Commodities" (2026).

### Why this source matters
- The report establishes the Strait of Hormuz as a critical energy-trade chokepoint.
- The volume and composition of flows help explain why a disruption could create a substantial supply shock to global energy markets.

### Visualization focus
- **Graph A:** Annual Hormuz oil flows, 2020–1H25. This chart shows total flows rising from 19.2 mb/d in 2020 to 20.9 mb/d in 1H 2025.
- **Graph B:** Stacked area chart of crude oil versus petroleum products. Crude remains the dominant flow, reinforcing that a blockade would most strongly affect crude supply and tanker demand.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Ensure visualization directory exists
os.makedirs("/visualizations", exist_ok=True)

# -------------------------------------------------------------------
# Graph A: Annual Hormuz oil flows, 2020 - 1H25
# -------------------------------------------------------------------
years = ["2020", "2021", "2022", "2023", "2024", "1H25"]
flows = [19.2, 19.5, 19.8, 20.1, 20.5, 20.9]

flow_df = pd.DataFrame({
    "Year": years,
    "Total Hormuz oil flow (mb/d)": flows
})

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(
    flow_df["Year"],
    flow_df["Total Hormuz oil flow (mb/d)"],
    marker="o",
    markersize=7,
    linewidth=2.5,
    color="#1f77b4"
)
ax.set_title("Hormuz Oil Flows, 2020 - 1H25", fontsize=14, weight="bold")
ax.set_xlabel("Year")
ax.set_ylabel("Oil flow (mb/d)")
ax.grid(True, linestyle="--", linewidth=0.6, alpha=0.5)
ax.set_ylim(18.5, 21.5)

for x, y in zip(flow_df["Year"], flow_df["Total Hormuz oil flow (mb/d)"]):
    ax.annotate(f"{y:.1f}", (x, y), textcoords="offset points", xytext=(0, 8), ha="center")

plt.tight_layout()
plt.savefig("C:/Users/jash/Desktop/foi_datas//visualizations/hormuz_oil_flows.png", dpi=300, bbox_inches="tight")
plt.close(fig)

# -------------------------------------------------------------------
# Graph B: Stacked area chart of crude vs petroleum products
# -------------------------------------------------------------------
stack_df = pd.DataFrame({
    "Year": ["2020", "2021", "2022", "2023", "2024", "1H25"],
    "Crude": [15.3, 15.5, 15.8, 16.1, 16.5, 16.9],
    "Petroleum products": [3.9, 4.0, 4.0, 4.0, 4.0, 4.0]
})

fig, ax = plt.subplots(figsize=(10, 6))

ax.stackplot(
    stack_df["Year"],
    stack_df["Crude"],
    stack_df["Petroleum products"],
    labels=["Crude", "Petroleum products"],
    alpha=0.75,
    colors=["#2ca02c", "#ff7f0e"]
)

ax.set_title("Hormuz Flow Composition: Crude vs Petroleum Products", fontsize=14, weight="bold")
ax.set_xlabel("Year")
ax.set_ylabel("Flow (mb/d)")
ax.legend(loc="upper left")
ax.grid(True, linestyle="--", linewidth=0.6, alpha=0.5)

plt.tight_layout()
plt.savefig("C:/Users/jash/Desktop/foi_datas/visualizations/hormuz_flow_composition.png", dpi=300, bbox_inches="tight")
plt.close(fig)

print("Saved visualizations to /visualizations")
print("- /visualizations/hormuz_oil_flows.png")
print("- /visualizations/hormuz_flow_composition.png")
```

## 2. Dallas Fed scenario analysis → oil-price and macro transmission

The Dallas Fed quarterly oil-market and macroeconomic model illustrates how a Strait of Hormuz supply disruption could affect oil prices and economic growth. The scenarios assume that close to 20% of global oil supply is removed and vary the closure duration. They are conditional model projections, not observed prices or forecasts of what will happen.

**Source:** Dallas Fed, Kilian, Plante & Richter (March 20, 2026).

### Why this source matters
- It links the physical supply shock in Graphs A and B to projected WTI prices and global real GDP growth under different reopening timelines.
- In all three scenarios, WTI reaches $98 per barrel in Q2 2026 and annualized global real GDP growth falls by 2.9 percentage points. After Q2, both paths depend on how long the closure lasts.
- With a one-quarter closure, WTI falls to $68 per barrel and GDP growth rises to 2.2 percentage points in Q3. With a three-quarter closure, WTI reaches $132 per barrel by Q4 and GDP growth remains negative at -1.1 percentage points.

### Visualization focus
- **Graph C:** Quarterly WTI price paths for one-, two-, and three-quarter closure scenarios. The paths diverge after the shared Q2 increase, illustrating how a longer disruption could prolong or intensify the price shock.
- **Graph D:** Annualized global real GDP growth paths, in percentage points, for the same closure scenarios. The chart compares the initial contraction with the different growth paths as the disruption continues or ends.

### Supplementary visualizations
These two Brent charts are additional context, not part of the six source-comparison graphs and not Dallas Fed model outputs. They use the Brent series supplied in the workbook; verify that series against EIA data before interpreting it as an observed market record.
- **Supplementary chart 1 — Brent price response:** A daily Brent price line chart with the disruption window shaded, showing price movements around the event. Saved as `brent_price_disruption.png`.
- **Supplementary chart 2 — Brent event study:** A bar chart comparing average Brent prices in the 30-day windows before, during, and after the disruption. The averages depend on the dates and observations available in the supplied series. Saved as `brent_event_study.png`.

## 3. CNBC article → equity-market outcomes

Use the CNBC article to examine the secondary-market outcome: whether tanker and energy-related equities outperformed the broad market during the disruption.

**Source:** CNBC article (add the article title, publication date, and URL to the final paper citation).

### Why this source matters
- Comparing tanker, oil-company, and broad-market returns helps assess which equities gained value over the selected period.
- The ~68% shipping-stock gain reported by CNBC is a reference claim, not a return calculated from the notebook. Verify its measurement date and basket definition before comparing it with the charts.
- These charts show market performance and do not, on their own, establish that transport scarcity caused the gains.

### Visualization focus
- **Graph E:** Year-to-date adjusted-price returns from the first common trading session in January 2026 through the latest common trading session. The chart includes FRO and DHT individually; equal-weight shipping (FRO, DHT), integrated oil-major (XOM, SHEL), and upstream (COP, EOG) baskets; their individual constituents; and SPY as the broad-market benchmark. Basket returns are the equal-weight average of constituent returns over the period.
- **Graph F:** Adjusted-price indices for FRO, DHT, XOM, SHEL, XLE, and SPY, each rebased to 100 on the first common January 2026 trading session.

Yahoo Finance adjusted closes are used for the charts. The ~68% CNBC figure is shown as a reference marker on Graph E; compare it with the calculated returns only after confirming that the date range and basket definition match.

```python
import json
import time
from datetime import datetime, timedelta, timezone
from pathlib import Path
from urllib.parse import urlencode
from urllib.request import Request, urlopen

# Download adjusted closes so dividends and splits are reflected in returns.
cnbc_tickers = ["FRO", "DHT", "XOM", "SHEL", "COP", "EOG", "SPY", "XLE"]
period_start = int(datetime(2026, 1, 1, tzinfo=timezone.utc).timestamp())
period_end = int((datetime.now(timezone.utc) + timedelta(days=1)).timestamp())
price_series = {}

for ticker in cnbc_tickers:
    params = urlencode({"period1": period_start, "period2": period_end, "interval": "1d", "events": "div,splits"})
    url = f"https://query1.finance.yahoo.com/v8/finance/chart/{ticker}?{params}"
    request = Request(url, headers={"User-Agent": "Mozilla/5.0"})
    with urlopen(request, timeout=30) as response:
        payload = json.load(response)

    chart = payload.get("chart", {})
    if chart.get("error") or not chart.get("result"):
        raise RuntimeError(f"Yahoo Finance returned no price history for {ticker}: {chart.get('error')}")
    result = chart["result"][0]
    timestamps = result.get("timestamp", [])
    adjusted = result.get("indicators", {}).get("adjclose", [{}])[0].get("adjclose", [])
    if not timestamps or len(timestamps) != len(adjusted):
        raise ValueError(f"Missing or incomplete adjusted-close history for {ticker}")

    series = pd.Series(
        adjusted,
        index=pd.to_datetime(timestamps, unit="s", utc=True).date,
        name=ticker,
        dtype="float64",
    )
    price_series[ticker] = series.dropna()
    time.sleep(0.2)

# Use common sessions so every return and index has the same comparison dates.
cnbc_prices = pd.concat(price_series.values(), axis=1, join="inner").dropna().sort_index()
if len(cnbc_prices) < 2:
    raise ValueError("At least two common trading sessions are required to calculate YTD returns")
baseline_date = cnbc_prices.index[0]
latest_date = cnbc_prices.index[-1]
cnbc_indexed = cnbc_prices.div(cnbc_prices.iloc[0]).mul(100)

baskets = {
    "Shipping basket (FRO, DHT)": ["FRO", "DHT"],
    "Oil majors basket (XOM, SHEL)": ["XOM", "SHEL"],
    "Upstream basket (COP, EOG)": ["COP", "EOG"],
}
graph_e_returns = (cnbc_indexed.iloc[-1] - 100).to_dict()
for label, members in baskets.items():
    graph_e_returns[label] = cnbc_indexed[members].mean(axis=1).iloc[-1] - 100

graph_e_order = [
    "FRO", "DHT", "Shipping basket (FRO, DHT)",
    "XOM", "SHEL", "Oil majors basket (XOM, SHEL)",
    "COP", "EOG", "Upstream basket (COP, EOG)", "SPY",
]
graph_e_values = pd.Series(graph_e_returns).reindex(graph_e_order).sort_values()
fig, ax = plt.subplots(figsize=(11, 7))
colors = ["#2ca02c" if value >= 0 else "#d62728" for value in graph_e_values.values]
bars = ax.barh(graph_e_values.index, graph_e_values.values, color=colors)
ax.axvline(0, color="black", linewidth=0.8)
ax.axvline(68, color="#1f77b4", linestyle="--", linewidth=1.5, label="CNBC-reported ~68% (verify period/basket)")
ax.set_title(f"YTD Adjusted-Price Returns: {baseline_date} to {latest_date}")
ax.set_xlabel("Total return (%)")
ax.grid(axis="x", linestyle="--", alpha=0.4)
ax.legend()
for bar, value in zip(bars, graph_e_values.values):
    offset = 3 if value >= 0 else -3
    ax.annotate(f"{value:.1f}%", (value, bar.get_y() + bar.get_height() / 2), xytext=(offset, 0), textcoords="offset points", va="center", ha="left" if value >= 0 else "right")
plt.tight_layout()
plt.savefig("C:/Users/jash/Desktop/foi_datas/visualizations/cnbc_ytd_returns.png", dpi=300, bbox_inches="tight")
plt.close(fig)

graph_f_tickers = ["FRO", "DHT", "XOM", "SHEL", "XLE", "SPY"]
fig, ax = plt.subplots(figsize=(11, 6))
for ticker in graph_f_tickers:
    ax.plot(cnbc_indexed.index, cnbc_indexed[ticker], linewidth=2, label=ticker)
ax.axhline(100, color="black", linewidth=0.8, alpha=0.7)
ax.set_title(f"Indexed Adjusted Prices (First Common January Session = 100): {baseline_date} to {latest_date}")
ax.set_xlabel("Date")
ax.set_ylabel("Indexed adjusted price (Jan 2026 = 100)")
ax.grid(True, linestyle="--", alpha=0.4)
ax.legend(ncol=3)
fig.autofmt_xdate()
plt.tight_layout()
plt.savefig("C:/Users/jash/Desktop/foi_datas/visualizations/cnbc_indexed_prices.png", dpi=300, bbox_inches="tight")
plt.close(fig)

print(f"Saved CNBC Graphs E and F for {baseline_date} through {latest_date}.")
print("YTD adjusted-price returns (%):")
print(graph_e_values.round(2).to_string())
```
